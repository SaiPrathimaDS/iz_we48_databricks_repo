### using dbutils.notebook.run 

1. this will invoke notebooke in diffrent instance / seprately / diffrent session  / isolate environemnet
2. var / fn we created in the child nb , we cant access in main notebook
3. timeout arg 
4. accept arguments and return value (string) to caller 

```python
dbutils.notebook.run("./child_dbutil_nb",60)
```

```python
%run ./child_dbutil_nb
```
