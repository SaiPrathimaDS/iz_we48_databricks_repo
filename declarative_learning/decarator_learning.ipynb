```python
from datetime import datetime
def log_fn_detail(fn):
    def wrapper():        
        print(f"INFO : {datetime.now()} function started executing .......")
        print("*"*25)
        fn()
        print("*"*25)
        print(f"INFO : {datetime.now()} function completed .............")

    return wrapper


# log_fn_detail takes a function as input and return a function output 
```

```python
@log_fn_detail
def print_detail():
    print("welcome to declrative programming")
    print("declrative style now available in community spark as well")
    print("This is the new feature ")
```
